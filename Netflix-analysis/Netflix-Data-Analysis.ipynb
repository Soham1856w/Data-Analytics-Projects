```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('netflix_titles.csv')
```

```python
# Top 10 countries.
top_countries=df['country'].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_countries.index,
    y=top_countries.values
)

plt.xticks(rotation=45)
plt.title('Top 10 Content producing Countries.')
plt.show()
```

### This Graph Shows us that which country has the highest content production rate.

```python
# Lets make release year trend graph

release_year = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))

sns.lineplot(
    x=release_year.index,
    y=release_year.values
)

plt.title("Netflix Content Release Trend")

plt.xlabel("Release Year")

plt.ylabel("Number of Titles")

plt.show()
```

```python
# lets analys content ratings.
plt.figure(figsize=(10,6))

sns.countplot(
    y='rating',
    data=df,
    order=df['rating'].value_counts().index
)
plt.title('Ratings Distribution on Netflix.')
plt.xlabel('Count')
plt.ylabel('Ratings')
plt.show()
```

### Insight:
TV-MA is the most common rating category on Netflix, showing strong focus on mature audiance.

```python
# Lets analys movie duration.
movies=df[df['type']=='Movie'].copy()
```

```python
movies['duration_int'] = movies['duration'].str.replace(' min', '')
```

```python
movies['duration_int'] = pd.to_numeric(
    movies['duration_int'],
    errors='coerce'
)
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    movies['duration_int'].dropna(),
    bins=30
)

plt.title("Movie Duration Distribution")

plt.xlabel("Duration in Minutes")

plt.ylabel("Number of Movies")

plt.show()
```

### Insight:
Most movies on Netflix are between 80 and 120 minutes long.

```python
# 10 genres
top_genres = df['listed_in'].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Genres on Netflix")

plt.xlabel("Count")

plt.show()
```

# Final Conclusion

- Netflix has significantly more Movies than TV Shows.
- The United States dominates Netflix content production.
- Netflix content releases increased rapidly after 2015.
- TV-MA is the most common content rating.
- Most movies are around 80 –120 minutes long.
